```python
from pathlib import Path
import os
import sys

root = next(path for path in [Path.cwd(), *Path.cwd().parents] if (path / 'config/paper.json').is_file())
os.chdir(root)
if str(root) not in sys.path:
    sys.path.insert(0, str(root))

import subprocess
output = Path('Figures/Generated')
output.mkdir(parents=True, exist_ok=True)
for name in ['Figure 01 Information Timeline', 'Figure 05 Workflow', 'Figure 06 Agent Roles']:
    subprocess.run(['node', 'FigureSources/render_svg.js', f'FigureSources/{name}.svg', str(output / f'{name}.png')], check=True)
```
