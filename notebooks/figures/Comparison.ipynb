```python
from pathlib import Path
import os
import sys

root = next(path for path in [Path.cwd(), *Path.cwd().parents] if (path / 'config/paper.json').is_file())
os.chdir(root)
if str(root) not in sys.path:
    sys.path.insert(0, str(root))

import json
import numpy as np
import pandas as pd
scores=pd.read_csv('FINDER/Official/daily_scores_all_markets.csv')
thresholds=pd.read_csv('FINDER/Official/daily_thresholds.csv')
frames=[]
for market,code in [('^GSPC','SP500'),('FTSE100','FTSE'),('GDAXI','DAX'),('CACT','CAC'),('STOXX','STOXX'),('CBU0.L','CBU0')]:
    frame=scores[scores.market.eq(market)].copy()
    frame['score']=frame.decision_supervisor
    supervisor=thresholds[thresholds.market.eq(market)&thresholds.agent.eq('decision_supervisor')][['target_date','threshold','prediction']]
    frame=frame.merge(supervisor.rename(columns={'threshold':'supervisor_threshold','prediction':'supervisor_prediction'}),on='target_date',validate='one_to_one')
    for model,key in [('IND','lstm_ind'),('NEWS','lstm_news')]:
        folder=Path('Benchmark/Official')/f'{code}_{model}'
        benchmark=pd.read_csv(folder/'test_predictions.csv')
        aligned=frame[['target_date','actual_label']].merge(benchmark,on='target_date',validate='one_to_one')
        assert len(aligned)==len(frame) and aligned.actual_label.eq(aligned.actual).all()
        frame=frame.merge(benchmark[['target_date','probability','prediction']].rename(columns={'probability':key,'prediction':key+'_prediction'}),on='target_date',validate='one_to_one')
        frame[key+'_threshold']=json.loads((folder/'metrics.json').read_text())['threshold']
    frames.append(frame)
joined=pd.concat(frames,ignore_index=True)
expected=pd.read_csv('FigureSources/daily_plot_data.csv')
joined=joined[expected.columns].sort_values(['market','target_date']).reset_index(drop=True)
expected=expected.sort_values(['market','target_date']).reset_index(drop=True)
pd.testing.assert_frame_equal(joined,expected,check_dtype=False,check_exact=False,rtol=1e-12,atol=1e-14)
output=Path('Figures/Generated')
output.mkdir(parents=True,exist_ok=True)
joined.to_csv(output/'daily_plot_data.csv',index=False)
```

```python
from pathlib import Path
import json
import hashlib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from matplotlib.ticker import MaxNLocator
plt.rcdefaults()
out = Path('Figures/Generated')
out.mkdir(parents=True, exist_ok=True)
# Run 02 Review Results.ipynb first to refresh the point estimates and intervals.
interval_path = Path('Results/Confidence Intervals/metrics_ci.csv')
if not interval_path.is_file():
    interval_path = Path('FINDER/Uncertainty/metrics_ci.csv')
intervals = pd.read_csv(interval_path)
assert intervals.confidence_level.eq(0.95).all()
assert intervals.method.eq('stationary_bootstrap_percentile').all()
model_keys = ['decision_supervisor', 'lstm_ind', 'lstm_news']
series_offset = 0.16
daily = pd.read_csv('FINDER/Official/daily_scores_all_markets.csv')
markets = [('^GSPC', 'SP500', 'S&P 500'), ('FTSE100', 'FTSE', 'FTSE 100'), ('GDAXI', 'DAX', 'DAX'), ('CACT', 'CAC', 'CACT'), ('STOXX', 'STOXX', 'STOXX'), ('CBU0.L', 'CBU0', 'CBU0.L')]
series_names = ['Decision Supervisor', 'LSTM-IND', 'LSTM-(IND+NEWS)']
series_colors = ['royalblue', '#555F6B', '#555F6B']
prevalence = daily.groupby('market').actual_label.mean()
plt.rcdefaults()
plt.rcParams['pdf.fonttype'] = 42
checks = []
for metric in ['auc', 'ap']:
    plt.figure(figsize=(5.3, 3.2))
    point_artists, interval_artists, interval_tables = [], [], []
    for j, (name, color, marker) in enumerate(zip(series_names, series_colors, ['o', 's', '^'])):
        table = intervals.loc[intervals.model.eq(model_keys[j]) & intervals.metric.eq('roc_auc' if metric == 'auc' else 'ap')].set_index('market').loc[[m[0] for m in markets]]
        assert np.isfinite(table[['estimate', 'ci_lower', 'ci_upper']]).all().all() and table.ci_lower.le(table.ci_upper).all()
        positions = np.arange(6) + (j - 1) * series_offset
        interval_artists.append(plt.errorbar((table.ci_lower + table.ci_upper) / 2, positions, xerr=(table.ci_upper - table.ci_lower) / 2, fmt='none', ecolor=color, elinewidth=0.8, capsize=2, capthick=0.8, alpha=0.4, zorder=2))
        point_artists.append(plt.scatter(table.estimate, positions, color=color, marker=marker, s=50, edgecolor='white', linewidth=0.65, zorder=3 + j))
        interval_tables.append(table)
    if metric == 'auc':
        plt.axvline(0.5, color='black', linestyle='-', linewidth=0.8)
        plt.xlim(0, 1)
        plt.xticks(np.arange(0, 1.01, 0.2), fontsize=12)
    else:
        upper = max(0.26, np.ceil((max(table.ci_upper.max() for table in interval_tables) + 0.01) / 0.05) * 0.05)
        plt.xlim(0, upper)
        plt.xticks(np.arange(0, upper + 0.001, 0.1 if upper > 0.3 else 0.05), fontsize=12)
        plt.scatter([prevalence[m[0]] for m in markets], np.arange(6), marker='|', s=240, color='black', linewidth=0.8, zorder=2)
    plt.yticks(np.arange(6), [m[2] for m in markets], fontsize=12)
    plt.ylim(5.45, -0.45)
    plt.xlabel('ROC-AUC (95% CI)' if metric == 'auc' else 'Average precision (95% CI)', fontsize=13)
    for y in np.arange(0.5, 5, 1):
        plt.axhline(y, linestyle='--', color='black', alpha=0.2, linewidth=0.7, zorder=0)
    plt.tight_layout()
    fig, ax = (plt.gcf(), plt.gca())
    fig.canvas.draw()
    renderer = fig.canvas.get_renderer()
    labels = [ax.xaxis.label, *ax.get_xticklabels(), *ax.get_yticklabels()]
    clipped = [label.get_text() for label in labels if not fig.bbox.contains(*label.get_window_extent(renderer).get_points()[0]) or not fig.bbox.contains(*label.get_window_extent(renderer).get_points()[1])]
    assert not clipped, clipped
    assert len(ax.get_yticks()) == 6 and all((spine.get_visible() for spine in ax.spines.values()))
    assert [collection.get_offsets().shape[0] for collection in point_artists] == [6, 6, 6]
    for j, (points, bars, table) in enumerate(zip(point_artists, interval_artists, interval_tables)):
        assert np.array_equal(points.get_offsets()[:, 0], table.estimate.to_numpy())
        assert points.get_alpha() in [None, 1] and bars.lines[2][0].get_alpha() == 0.4
        segments = np.array(bars.lines[2][0].get_segments())
        assert np.allclose(segments[:, :, 0], table[['ci_lower', 'ci_upper']].to_numpy())
        assert np.allclose(segments[:, :, 1], (np.arange(6) + (j - 1) * series_offset)[:, None])
        assert table.ci_lower.ge(ax.get_xlim()[0]).all() and table.ci_upper.le(ax.get_xlim()[1]).all()
    checks.append({'metric': metric, 'dimensions_inches': [5.3, 3.2], 'font': plt.rcParams['font.sans-serif'][0], 'visible_spines': list(ax.spines), 'model_y_offsets': [-series_offset, 0, series_offset], 'confidence_level': 0.95, 'interval_method': 'stationary_bootstrap_percentile', 'horizontal_separator_alpha': 0.2, 'xlim': list(ax.get_xlim()), 'market_spacing_points': (ax.transData.transform((0, 1))[1] - ax.transData.transform((0, 0))[1]) * 72 / fig.dpi, 'references': 0.5 if metric == 'auc' else {m[2]: float(prevalence[m[0]]) for m in markets}, 'clipped_labels': clipped, 'plotted_values_and_bounds_equal_ci_table': True})
    filename = 'Figure 08 A ROC AUC.png' if metric == 'auc' else 'Figure 08 B Average Precision.png'
    plt.savefig(out / filename, dpi=360, facecolor='white')
    plt.show()
    plt.close()
```
